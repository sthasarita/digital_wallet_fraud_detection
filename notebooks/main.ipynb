```python
# step 2 : Prepare a separate modelling dataset 
# keep only columns we want to investigate initially
model_df = df[["step","type", "amount", "oldbalanceOrg","isFraud" ]].copy()

# check the new dataset's size
print("Shape:", model_df.shape)
# Display the first five rows
print(model_df.head())

# check whether any values are missing
print("Missing values:")
print(model_df.isnull().sum())

# check the fraud label distribution 
print("Fraud label counts:")
print(model_df["isFraud"].value_counts())
```

```python
#Check the earliest and latest simulation steps
print("Earliest step:", model_df["step"].min())
print("Latest step:", model_df["step"].max())

# See how many transactions occur at each step
print("Unique steps:", model_df["step"].nunique())

# inspect transactions counts across the timeline
print(model_df.groupby("step").size().describe())
```

```python
# Find the boundary between training and testing periods 
split_step = int(model_df["step"].max() * 0.8)

print("Split step:", split_step)

# separate the earlier and later transactions
train_data = model_df[model_df["step"] <= split_step]
test_data = model_df[model_df["step"] > split_step]

print("Training data shape:", train_data.shape)
print("Testing data shape:", test_data.shape)

print("Fraud cases in training data:")
print(train_data["isFraud"].value_counts())

print("Fraud cases in testing data:")
print(test_data["isFraud"].value_counts())

print("Fraud rate in training data:")
print(train_data["isFraud"].mean() * 100, "%")

print("Fraud rate in testing data:")
print(test_data["isFraud"].mean() * 100, "%")
```

```python
# Input features for the model
features = ["amount", "oldbalanceOrg", "type"]

# The answer the model must learn to predict
target = "isFraud"

# Separate inputs(X) and answes(y)
X_train = train_data[features]
y_train = train_data[target]

X_test = test_data[features]
y_test = test_data[target]

print("Training features:", X_train.shape)
print("Training target:", y_train.shape)

print("Testing features", X_test.shape)
print("Testing target:", y_test.shape)

print("Sample training features:")
print(X_train.head())
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Convert the text column "type" into numerical columns 
preprocessor = ColumnTransformer(
    transformers = [
        ("type_encoder", OneHotEncoder(handle_unknown = "ignore"),
        ["type"])
    ],
    remainder = "passthrough"
)

#learn the encoding from training data and transform it 
X_train_encoded = preprocessor.fit_transform(X_train)

# Apply the same encodeing to testing data 
X_test_encoded = preprocessor.transform(X_test)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)

print("Encoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python
# experiment with different thresholds
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.3, 0.5, 0.7, 0.9]
results = []

for threshold in thresholds:
    # convert probabilitiees into predictions
    predictions = (y_probability >= threshold).astype(int)

    #calculate evaluation metrics 
    results.append(
        {
            "Threshold" : threshold,
            "Precision" : precision_score(
                y_test,predictions, zero_division = 0
            ),
            "Recall" : recall_score(
                y_test, predictions, zero_division = 0
            ),
            "F1-score" : f1_score(
                y_test, predictions, zero_division = 0
            ),
            "Flagged transactions" : int(predictions.sum())
        }
    )

    results_df = pd.DataFrame(results)
    print(results_df.round(3).to_string(index = False))
```

```python
def prepare_autoencoder_data(data, preprocessor, scaler):
    #Encode categorical features
    encoded = preprocessor.transform(data)

    #Convert to dense float32
    if sparse.issparse(encoded):
        encoded = encoded.toarray().astype(np.float32)
    else:
        encoded = np.asarray(encoded, dtype = np.float32)

    # apply log transformation to monetary features
    encoded[:, 5] = np.log1p(encoded[:, 5])
    encoded[:, 6] = np.log1p(encoded[:, 6])

    #apply the previously fitted scaler
    encoded[:, 5:7] = scaler.transform(encoded[:, 5:7])

    return encoded

#prepare validation data
X_val_autoencoder = prepare_autoencoder_data(
    X_val, preprocessor_val, scaler
)

#prepare test data
X_test_autoencoder = prepare_autoencoder_data(
    X_test, preprocessor_val, scaler
)

print("Validation shape:", X_val_autoencoder.shape)
print("Test Shape:", X_test_autoencoder.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Reproducibility
tf.keras.utils.set_random_seed(42)

# Build the complete autoencoder
autoencoder = keras.Sequential([
    keras.Input(shape=(12,)),

    # Encoder
    layers.Dense(8, activation='relu'),
    layers.Dense(4, activation='relu'),

    # Decoder
    layers.Dense(8, activation='relu'),
    layers.Dense(12, activation='linear')
])

# Compile the model
autoencoder.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='mse'
)

# Display architecture
autoencoder.summary()

# Early stopping
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# Train the autoencoder
history = autoencoder.fit(
    X_train_normal_scaled,
    X_train_normal_scaled,
    epochs=30,
    batch_size=2048,
    validation_data=(X_val_scaled, X_val_scaled),
    callbacks=[early_stopping],
    shuffle=False
)
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    average_precision_score,
    classification_report
)

# --------------------------------------------------
# 1. Calculate precision, recall and thresholds
#    using validation data only
# --------------------------------------------------

val_precision, val_recall, val_thresholds = precision_recall_curve(
    y_val_ae3,
    val_errors
)

# --------------------------------------------------
# 2. Select thresholds for target recall levels
# --------------------------------------------------

target_recalls = [0.70, 0.80]

results = []

for target_recall in target_recalls:

    # Find thresholds that achieve at least the target recall
    eligible_indices = np.where(
        val_recall[:-1] >= target_recall
    )[0]

    if len(eligible_indices) == 0:
        print(f"No threshold achieves {target_recall:.0%} recall.")
        continue

    # Select the highest threshold that still meets the target recall
    best_index = eligible_indices[-1]
    threshold = val_thresholds[best_index]

    # --------------------------------------------------
    # 3. Apply the selected threshold to test data
    # --------------------------------------------------

    test_predictions = (test_errors > threshold).astype(int)

    precision = precision_score(y_test_ae3, test_predictions)
    recall = recall_score(y_test_ae3, test_predictions)
    f1 = f1_score(y_test_ae3, test_predictions)

    cm = confusion_matrix(y_test_ae3, test_predictions)

    tn, fp, fn, tp = cm.ravel()

    results.append({
        "Target Recall": f"{target_recall:.0%}",
        "Threshold": threshold,
        "Test Precision": precision,
        "Test Recall": recall,
        "Test F1": f1,
        "True Positives": tp,
        "False Positives": fp,
        "False Negatives": fn,
        "True Negatives": tn
    })

    print(f"\n{'=' * 55}")
    print(f"TARGET RECALL: {target_recall:.0%}")
    print(f"{'=' * 55}")

    print(f"Selected threshold: {threshold:.6f}")
    print(f"Test precision:    {precision:.4f}")
    print(f"Test recall:       {recall:.4f}")
    print(f"Test F1-score:     {f1:.4f}")

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(classification_report(
        y_test_ae3,
        test_predictions,
        digits=4
    ))

# --------------------------------------------------
# 4. Compare both operating points
# --------------------------------------------------

results_df = pd.DataFrame(results)

print("\n\nFINAL COMPARISON")
display(results_df)
```
